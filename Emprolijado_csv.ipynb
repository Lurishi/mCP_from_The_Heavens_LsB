```python
cosine_values = [-1,-0.8,-0.6,-0.4,-0.2,0,0.2,0.4,0.6,0.8,1]
factor = [1,0.878,0.765,0.651,0.511,0.210,0.511,0.651,0.765,0.878,1]
list = []
for i in range(len(cosine_values)):
    df_cosine = df2.copy()
    df_cosine['cos_theta'] = cosine_values[i]
    df_cosine['flux_differential'] = df_cosine['flux_differential']*factor[i]
    list.append(df_cosine)
df_cosine_merged = pd.concat(list, ignore_index=True)
df_cosine_merged = df_cosine_merged.rename(columns={"flux_differential": "dphi_dEdCth(1/GeV/s/cm2/sr)"}) 
df_cosine_merged = df_cosine_merged.rename(columns={"Energy": "Energy(GeV)"})

df_cosine_merged.to_csv('Data/mCP_PB_differential_flux_surface.csv', index=False)
```
